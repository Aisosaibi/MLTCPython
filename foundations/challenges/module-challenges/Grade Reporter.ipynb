<details>
  <summary style="color: grey;">Hint 1: Getting a Score and Basic Grade</summary>

  **Goal:** Ask the user for a number (their score) and then figure out if it's an 'A', 'B', 'C', 'D', or 'F'.

  1.  **Ask for Input:** Use `input()` to ask the user to type in their score. This will be a text (string).

  2.  **Turn into a Number:** Convert the text score into a whole number (an integer) using `int()`. For example, if they type '85', `int('85')` becomes the number 85.

  3.  **Check the Grade:** Use `if`, `elif` (short for 'else if'), and `else` statements to compare the score with the ranges for A, B, C, D, and F. Start with the highest grades first (e.g., if score is 90 or more, it's an 'A').

  4.  **Show the Grade:** Use `print()` to show the user their letter grade.
</details>



<details>
  <summary style="color: grey;">Hint 2: Keep Asking and Know When to Stop</summary>

  **Goal:** Make the program keep asking for scores until the user types 'quit'. Also, make sure the score is a normal test score (between 0 and 100).

  1.  **Endless Loop:** Put your grading code inside a `while True:` loop. This makes the program repeat forever until you tell it to stop.

  2.  **Stop When 'quit':** Inside the loop, after asking for input, check if what the user typed is the word 'quit' (you might want to check for 'Quit' or 'QUIT' too, using `.lower()` to make everything lowercase for easy comparison). If it's 'quit', use `break` to stop the loop.

  3.  **Valid Score Check:** Before calculating the grade, check if the number the user entered is actually between 0 and 100. If it's not, print a message saying it's invalid and use `continue` to skip the rest of the loop and ask for input again.
</details>



<details>
  <summary style="color: grey;">Hint 3: Handle Tricky Inputs (Preventing Crashes!)</summary>

  **Goal:** Make sure your program doesn't crash if the user types something unexpected, like letters instead of numbers.

  1.  **The Problem:** Remember when you used `int()` to turn text into a number? What if someone types 'hello' instead of '85'? Your program will stop working with an error called `ValueError`.

  2.  **The Solution: `try` and `except`**
      Python has a special way to "try" doing something and "catch" errors if they happen. It's like saying: "*Try* to do this, but if an *error* happens, *except* that error and do something else instead."

      Here's a super simple example of how `try` and `except` work:
      ```python
      try:
          # Try to do something that might cause an error
          number = int(input("Enter a number: "))
          print(f"You entered: {number}")
      except ValueError:
          # If a ValueError happens in the 'try' part, do this instead
          print("That's not a valid number!")
      ```

  3.  **Apply to Your Program:** Now, you can put the part of your code where you convert the input to an `int()` (and potentially other things that might go wrong) inside a `try` block. If the user types something that can't be turned into a number, your `except ValueError:` block will catch it, print a friendly error message, and then you can use `continue` to ask for input again without the program crashing!
</details>
